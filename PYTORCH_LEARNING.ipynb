```python
import torch

x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

y.backward()

print("Output:", y.item())
print("Gradient:", x.grad.item())
```

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = 3 * x**2 + 2 * x + 1

y.backward()

print("Output:", y.item())
print("Gradient:", x.grad.item())
```

```python
x = torch.tensor(4.0, requires_grad=True)
y = x ** 2
y.backward()

print(x.grad.item())
print(x.grad)
```

```python
import torch

x = torch.tensor(3.0, requires_grad=True)

a = x * 2
y = a.detach() * x

y.backward()

print(y.item())
print(x.grad.item())
```

```python
import torch

x = torch.tensor(3.0, requires_grad=True)

y = x ** 2
z = y.detach()

print(y.requires_grad)
print(z.requires_grad)
```

```python
import torch

x = torch.tensor(3.0, requires_grad=True)

with torch.no_grad():
    y = x ** 2

print(y.requires_grad)
```
