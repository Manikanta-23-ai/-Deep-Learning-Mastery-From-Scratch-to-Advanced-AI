Backpropagation helps a neural network figure out how much each weight contributed to its error.


```python
x = 2.0
target = 10.0
weight = 3.0

prediction = weight * x
loss = (prediction - target) ** 2

gradient = 2 * (prediction - target) * x

print("Prediction:", prediction)
print("Loss:", loss)
print("Gradient:", gradient)
```

```python
x, target = 2.0, 10.0
weight, bias = 1.0, 0.0
lr = 0.01

for step in range(100):
    raw = weight * x + bias          # Neuron
    pred = max(0, raw)                # ReLU
    error = pred - target
    loss = error ** 2
    dw = 2 * error * x if raw > 0 else 0
    db = 2 * error if raw > 0 else 0
    weight -= lr * dw
    bias -= lr * db

print("Prediction:", round(max(0, weight*x + bias), 2))
print("Loss:", round(loss, 4))
```
