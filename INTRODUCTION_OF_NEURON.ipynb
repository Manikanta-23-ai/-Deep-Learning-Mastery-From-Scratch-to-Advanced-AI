```python
import numpy as np

X = np.array([2.0, 3.0])
W = np.array([0.5, 0.2])
b = 1.0

raw_output = np.dot(X, W) + b
final_output = np.maximum(0, raw_output)

print("Raw output:", raw_output)
print("After ReLU:", final_output)
```
