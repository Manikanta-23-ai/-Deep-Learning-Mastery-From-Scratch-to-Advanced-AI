```python
import numpy as np

a = np.array([2, 4, 6])
b = np.array([1, 3, 5])

print("Addition:", a + b)
print("Multiplication:", a * b)
print("Scalar multiplication:", a * 2)
print("Dot product:", np.dot(a, b))
```

```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

print("A shape:", A.shape)
print("B shape:", B.shape)
print("Matrix product:\n", A @ B)
```

```python
import numpy as np

X = np.array([
    [2, 3],
    [4, 5],
    [6, 7]
])

W = np.array([
    [0.5, 0.2, 0.1],
    [0.3, 0.4, 0.6]
])

Y = X @ W

print(Y)
print("X shape:", X.shape)
print("W shape:", W.shape)
print("Y shape:", Y.shape)
```

```python
b = np.array([0.1, -0.2, 0.3])
Y = X @ W + b * b
```
