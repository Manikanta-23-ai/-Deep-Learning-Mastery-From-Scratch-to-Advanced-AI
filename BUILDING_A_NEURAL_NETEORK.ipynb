```python
import torch
import torch.nn as nn

# 1. Training examples
X = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [4.0], [6.0], [8.0]])

# 2. Create the model
model = nn.Linear(1, 1)

# 3. Set loss and optimizer
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# 4. Train the model
for i in range(500):
    prediction = model(X)
    loss = loss_fn(prediction, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

# 5. Test the model
test = torch.tensor([[5.0]])
result = model(test)

print("Prediction:", result.item())
```

```python
import torch
import torch.nn as nn

# 1. Training data
X = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [4.0], [6.0], [8.0]])

# 2. Create a neural network with one neuron
model = nn.Linear(1, 1)

# 3. Choose loss function and optimizer
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# 4. Training loop
for epoch in range(500):
    prediction = model(X)             # Forward pass
    loss = loss_fn(prediction, y)     # Calculate error

    optimizer.zero_grad()              # Clear old gradients
    loss.backward()                    # Backpropagation
    optimizer.step()                   # Update parameters

# 5. Test the trained model
with torch.no_grad():
    result = model(torch.tensor([[5.0]]))

print("Prediction:", result.item())
print("Final loss:", loss.item())
print("Learned weight:", model.weight.item())
print("Learned bias:", model.bias.item())
```
