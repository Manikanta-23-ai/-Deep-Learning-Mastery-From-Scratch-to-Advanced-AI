```python
import torch
import torch.nn as nn

# 1. Training examples
X = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [4.0], [6.0], [8.0]])

# 2. Create the model
model = nn.Linear(1, 1)

# 3. Set loss and optimizer
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# 4. Train the model
for i in range(500):
    prediction = model(X)
    loss = loss_fn(prediction, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

# 5. Test the model
test = torch.tensor([[5.0]])
result = model(test)

print("Prediction:", result.item())
```

```python
import torch
import torch.nn as nn

# Create a neural network
model = nn.Sequential(
    nn.Linear(2, 4),
    nn.ReLU(),
    nn.Linear(4, 1)
)

# Give the model two examples
X = torch.tensor([
    [1.0, 2.0],
    [3.0, 4.0]
])

# Get the output
output = model(X)

print(output)
print(output.shape)
```

```python
import torch
import torch.nn as nn

relu = nn.ReLU()

x = torch.tensor([-3.0, -1.0, 0.0, 2.0, 5.0])

print(relu(x))
```

```python
sigmoid = nn.Sigmoid()

x = torch.tensor([-5.0, 0.0, 5.0])

print(sigmoid(x))
```
